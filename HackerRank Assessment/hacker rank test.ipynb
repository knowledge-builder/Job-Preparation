```python
import pandas as pd
import os
def df_filter(ids_list, names_list):
    df = pd.DataFrame({
        'id': ids_list,
        'name': names_list
    })
    
    df = df[
        ~df['name'].str.contains('a') &
        ~df['id'].str.contains('a')
    ]
    
    return len(df)
ids_list = input().split(" ")
names_list = input().split(" ")
res = df_filter(ids_list, names_list)
print(res)
```

```python
import pandas as pd
import os

def generate_df(salary_list, country_list):
    df = pd.DataFrame({
        'salary': [float(x) for x in salary_list],
        'country': country_list
    })

    # group salaries by country and calculate average salary for each group
    res = df.groupby('country')['salary'].mean().sort_values(ascending=False)

    # no need to change the output format
    return ' '.join(res.astype(str))

salary_list = input().split(" ")
country_list = input().split(" ")

res = generate_df(salary_list, country_list)
print(res)
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys



#
# Complete the 'stop_words' function below.
#
# The function is expected to return a STRING_ARRAY.
# The function accepts following parameters:
#  1. STRING text
#  2. INTEGER k
#

def stopWords(text, k):
    # Write your code here
    words = text.split()
    
    freq = {}
    
    # count frequencies
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    
    result = []
    seen = set()
    
    for word in words:
        if freq[word] >= k and word not in seen:
            result.append(word)
            seen.add(word)
    
    return result

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    text = input()

    k = int(input().strip())

    result = stopWords(text, k)

    fptr.write('\n'.join(result))
    fptr.write('\n')

    fptr.close()
```

```python


WITH daily AS (
    SELECT
        visited_on,
        SUM(amount) amount
    FROM customers
    GROUP BY visited_on
)

SELECT
    visited_on,
    amount,
    ROUND(
        AVG(amount) OVER (
            ORDER BY visited_on
            ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
        ),
        4
    ) avg_amount
FROM daily
ORDER BY visited_on

```
